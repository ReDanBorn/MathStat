```python
import pandas as pd
```

Получены данные статистического наблюдения - абсолютные величины заболеваемости в районе N за 2020 год.

1. Число детей, проживающих в районе N в 2020 году
 (all): 3500 человек

   Из них: *мальчики* - 1700 чел.,  *девочки* - 1800 чел.
   
2. Число болевших детей на протяжении 2020 года (BD): 2900 человек

   Из них: *мальчики* - 1400 чел., *девочки* - 1500 чел.

3. Общее число зарегистрированных заболеваний у детей за 2020 год (ZZ): 3820 случаев.

   Из них: *мальчики* - 1900 случаев, *девочки* - 1920 случаев.


4. Число заболеваний скарлатиной у детей за 2020 год (ZS): 66 случаев.

   Из них: *мальчики* - 35 случаев, *девочки* - 31 случай.

Задание

1. Вычислить показатели числа больных лиц, частоту случаев всех заболеваний и случаев скарлатиной (интенсивные величины) на 1000 детей

2. Построить статистическую таблицу, содержащую интенсивные показатели

3. Вычислить показатели доли  (удельного веса) мальчиков и девочек среди детей района и долю заболеваний скарлатиной во всех заболеваниях детей, а также среди мальчиков и девочек (экстенсивные)

4. Изобразить эксенсивные показатели секторной и внутристолбиковой диаграммами


Создадим таблицу с данными задачи

```python
df = pd.DataFrame(data={'all':[1700,1800,3500],
                         'BD':[1400,1500,2900],
                         'ZZ' :[1900,1920,3820],
                         'ZS':[35,31,66]}, index=['мальчики','девочки','всего'])
df.head()
```

1. Вычислить показатели числа больных лиц, частоту случаев всех заболеваний и случаев скарлатиной (интенсивные величины) на 1000 детей

а) Показатели числа больных лиц

```python
p_BD=(df['BD']/df['all'])*1000
p_BD
```

б) Частота случаев всех заболеваний

```python
p_ZZ = round((df['ZZ']/df['all'])*1000, 1)# ДОПИШИ КОД - СДЕЛАНО
p_ZZ
```

в) Частота случаев заболевания скарлатиной

```python
p_ZS = round((df['ZS']/df['all'])*1000, 1)# ДОПИШИ КОД - СДЕЛАНО
p_ZS
```

2. Построить статистическую таблицу, содержащую интенсивные показатели

```python
df1 = pd.DataFrame(data={'Число болевших на 1000 детей': p_BD, # ДОПИШИ КОД - СДЕЛАНО
                         'Число случаев заболеваний на 1000 детей': p_ZZ, # ДОПИШИ КОД - СДЕЛАНО
                         'Число случаев скарлатины на 1000 детей': p_ZS}, # ДОПИШИ КОД - СДЕЛАНО
                   index=['мальчики','девочки','всего'])
df1.index.name='Пол'
df1.head()
```

3.Вычислить показатели доли (удельного веса) мальчиков и девочек среди детей района и долю заболеваний скарлатиной во всех заболеваниях детей, а также среди мальчиков и девочек (экстенсивные)

```python
p_1=round((df.loc['мальчики','all']/df.loc['всего','all'])*100,1)
p_1
```

```python
p_2 = round((df.loc['девочки','all']/df.loc['всего','all'])*100, 1) # ДОПИШИ КОД - СДЕЛАНО
p_2
```

```python
p_ud_ves_skar = round((df.loc['всего','ZS']/df.loc['всего','ZZ'])*100, 1) # ДОПИШИ КОД - СДЕЛАНО
p_ud_ves_skar
```

```python
p_ud_ves_skar_b = round((df.loc['мальчики','ZS']/df.loc['мальчики','ZZ'])*100, 1) # ДОПИШИ КОД - СДЕЛАНО
p_ud_ves_skar_b
```

```python
p_ud_ves_skar_g = round((df.loc['девочки','ZS']/df.loc['девочки','ZZ'])*100, 1) # ДОПИШИ КОД - СДЕЛАНО
p_ud_ves_skar_g
```

4. Изобразить экстенсивные показатели секторной и внутристолбиковой диаграммами

```python
import matplotlib.pyplot as plt
labels = 'Мальчики', 'Девочки'
sizes = [p_1, p_2] # ДОПИШИ КОД - СДЕЛАНО
fig1, ax1 = plt.subplots()
ax1.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.title('Доля мальчиков и девочек, проживающих в районе N. Секторная диаграмма')
plt.show()
```

```python
labels = ['']

fig, ax = plt.subplots()

ax.bar(labels, p_1, linewidth=3, label='Мальчики', color = 'Blue')
ax.bar(labels, p_2, linewidth=3, bottom=p_1,# ДОПИШИ КОД - СДЕЛАНО
      label='Девочки', color = 'Red' )

ax.set_ylabel('Соотношение, в %')
first_legend = ax.legend(loc='upper right')
ax.add_artist(first_legend)
ax.legend(loc='center', labels = (p_1, p_2))# ДОПИШИ КОД - СДЕЛАНО
plt.title('Доля мальчиков и девочек, проживающих в районе N. Внутристолбиковая диаграмма')
plt.show()
```

# Задача 2.

Получены данные статистического. наблюдения: среднегодовая численность жителей города N. в 2020. году составляла 300 тыс. человек. За год зарегистрировано 216 850 случаев заболеваний, в том числе по группам: заболевания органов дыхания - 92 000 случаев, болезни органов кровообращения - 82 000 случаев, травмы - 16500 случаев, новообразования - 16 000 случаев, инфекционные болезни - 9 000 случаев, прочие заболевания - 1350.

**Задание**

Составьте статистическую таблицу, вычислите интенсивные показатели частоты и экстенсивные показатели структуры заболеваемости жителей города N за 2020 год. Отобразите эти показатели соответствующими диаграммами. Сделайте вывод.

```python
df2 = pd.DataFrame(data={'a':[92000,82000,16500,16000,9000,1350],
                         }, index=['Органов дыхания','Органов кровообращения','Травмы', 'Новообразования', 'Инфекционные болезни', 'Прочие'])
df2.index.name='Группы заболеваний'
df2
```

Интенсивные показатели частоты

```python
N = 300000  # среднегодовая численность жителей
df2['Интенсивный показатель на 1000 жителей'] = round((df2['a']/N)*1000, 1)
df2
```

Экстенсивные показатели структуры заболеваемости

```python
total_cases = df2['a'].sum()
df2['Экстенсивный показатель, %'] = round((df2['a']/total_cases)*100, 1)
df2
```

Диаграммы

```python
import matplotlib.pyplot as plt

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# --- Секторная диаграмма ---
labels = df2.index
sizes = df2['Экстенсивный показатель, %']
ax1.pie(sizes, autopct='%1.1f%%')
ax1.legend(labels, loc='center left', bbox_to_anchor=(1, 0.5), fontsize=9)
ax1.set_title('Структура заболеваемости жителей\nгорода N за 2020 год. Секторная диаграмма')

# --- Столбиковая диаграмма ---
ax2.bar(df2.index, df2['Интенсивный показатель на 1000 жителей'], color='teal')
ax2.set_ylabel('Число случаев на 1000 жителей')
ax2.set_title('Частота заболеваемости по группам болезней\n(на 1000 жителей). Столбиковая диаграмма')
ax2.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```
